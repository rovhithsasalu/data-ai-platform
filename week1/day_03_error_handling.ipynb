## 3. else and finally

else runs only if the try block completed without an exception. finally runs whether the operation succeeded or failed, often for cleanup.

```python
def parse_count(text):
    try:
        count = int(text)
    except ValueError:
        print("That was not an integer")
    else:
        print("Parsed:", count)
    finally:
        print("Parse attempt finished")

parse_count("12")
parse_count("twelve")
```

## 4. Raise an exception

Use raise to report that a function cannot accept an input. Choose an exception type and message that describe the problem.

```python
def positive_amount(text):
    try:
        amount = float(text)
    except ValueError as error:
        raise ValueError("amount must be numeric") from error

    if amount < 0:
        raise ValueError("amount must not be negative")

    return amount

print(positive_amount("12.5"))
```

## 5. Custom exception

Define a custom exception by subclassing Exception. Use one when it gives a clearer domain-specific error than a built-in exception.

```python
class InvalidOrderError(Exception):
    pass

def require_customer(order):
    if not order.get("customer_id"):
        raise InvalidOrderError("customer_id is required")
    return order

print(require_customer({"customer_id": "C001"}))
```
