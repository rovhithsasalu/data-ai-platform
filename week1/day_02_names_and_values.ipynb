## 1. Names and objects

A name is a label bound to an object. In `x = 10`, Python evaluates `10` and binds `x` to that object. A name is not a box that permanently contains a value. Reassigning `x` binds that name elsewhere; it does not change another name's binding.

```python
x = 10
y = x
x = 20
print(x, y)
```

## 2. Equality and identity

`==` asks whether values compare equal. `is` asks whether two names refer to the exact same object. Use `is None` to check for `None`; use `==` for ordinary value comparisons. Don't rely on object identity for numbers or strings.

```python
a = [1, 2]
b = a
c = [1, 2]
print(a == b, a is b)
print(a == c, a is c)
```

## 3. Mutable and immutable objects

A mutable object can change in place; lists and dictionaries are common examples. Strings, integers, and tuples are immutable. A tuple cannot have one of its slots rebound, but an object stored inside the tuple may itself be mutable.

```python
items = ["pen", "book"]
items.append("cup")  # mutates this list object
print(items)

label = "data"
label = label + "base"  # makes a new string and rebinds label
print(label)
```

## 4. Aliasing: two names, one mutable object

Assignment does not copy a list. After `second = first`, both names refer to the same list. A mutation through either name is visible through both. Use `.copy()` or another explicit copying method when you need an independent outer list.

```python
first = [10, 20]
second = first
second.append(30)
print(first)
print(first is second)

independent = first.copy()
independent.append(40)
print(first, independent)
print(first is independent)
```

## 5. Nested containers and shallow copies

Copying an outer list or dictionary is usually a shallow copy: the outer container is new, but nested mutable values can still be shared. List repetition also repeats references to an inner object; it does not recursively copy it.

```python
shared_rows = [[0, 0]] * 2
shared_rows[0][0] = 1
print(shared_rows)

separate_rows = [[0, 0] for _ in range(2)]
separate_rows[0][0] = 1
print(separate_rows)
```

## 6. Function arguments are new local names

When a function is called, its parameter name is bound to the argument object. The object is not automatically copied. Mutating a mutable argument is visible to the caller; rebinding the parameter name is local to the function.

```python
def add_tag(tags):
    tags.append("new")

def replace_tags(tags):
    tags = ["replacement"]
    return tags

labels = ["existing"]
add_tag(labels)
print(labels)
returned = replace_tags(labels)
print(labels)
print(returned)
```

## 7. Mutable default arguments

Default argument expressions are evaluated once when Python creates the function, not again on every call. A mutable default can therefore retain changes across calls. A common pattern is to use `None`, then create a new list inside the function.

```python
def add_item(item, bucket=[]):
    bucket.append(item)
    return bucket

print(add_item("a"))
print(add_item("b"))
print(add_item("c", []))
```

## 8. Augmented assignment and tuples containing lists

Augmented assignment such as `+=` can mutate an object in place, depending on its type. For a list inside a tuple, the list may be extended first; Python then attempts to assign the result back to the tuple slot, which is immutable. Predict both the error and the final tuple state.

```python
record = ("order", [1, 2])
try:
    record[1] += [3]
except TypeError as error:
    print(type(error).__name__)
print(record)
```

## 9. Closures and late binding

A nested function can refer to a name from an enclosing function. A closure keeps access to that binding. In a loop, lambdas that refer to the loop variable look up the variable when called; a default argument can capture the current value when the lambda is created.

```python
functions = [lambda: i for i in range(3)]
print([function() for function in functions])

functions = [lambda i=i: i for i in range(3)]
print([function() for function in functions])
```
