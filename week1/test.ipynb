```python
from collections import defaultdict
groups=defaultdict(int)
nums=[3,1,1,2,2,1]
for n in nums:
    groups[n]+=1
print(groups)
```

```python
buckets=[[] for _ in range(len(nums)+1)]
```

```python
print(buckets)
```

```python
for k,v in groups.items():
    buckets[k].append(v)

    
```

```python
for bucket in buckets:
    #print(buckets.index(bucket))
    if bucket==0:
        continue
    for v in bucket:
        print(v)
```
