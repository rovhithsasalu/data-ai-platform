First explain how `ROW_NUMBER()`, `RANK()`, and `DENSE_RANK()` handle tied values. Then write a query that returns the top two **distinct salary ranks** within each department, including all employees tied at either rank.

**Tie explanation:**

> Row Number return sequential numbers 1,2,3
rank, repeats same rank for tie and skip next number
dense rank, repeats same rank for tie but continue with next number without skipping

```python
-- Write your SQL query below.
-- Use DENSE_RANK() partitioned 
department, then keep ranks <= 2.
select * from (
select emp.*,
DENSE_RANK() over (partition by departemtn order by salary desc) as rnk)
from employees emp) where rnk <= 2
```
