### `range` and `while`

`range(3)` produces the integers starting at zero and stopping before three. A `while` loop repeats as long as its condition is true. Be sure something in the loop changes the condition, or it may never stop.

````python
for number in range(3):
    print(number)

step = 0
while step < 3:
    print(step)
    step += 1
````

## 6. Functions, parameters, and return values (15 min)

`def` defines a function. Parameters are names a function receives. `return` sends a result back to the caller; `print` displays something but does not send that value back to your code.

````python
def double(number):
    result = number * 2
    return result

answer = double(5)
print(answer)
````

### Positional and keyword arguments

A parameter can have a default value. If you leave it out when calling the function, Python uses that default. You can also name an argument explicitly, as in `quantity=2`.

````python
def describe(item, quantity=1):
    return f"{quantity} x {item}"

print(describe("book"))
print(describe("pen", 3))
print(describe("mug", quantity=2))
````
