## 1. Modules and imports

A module is a Python source file, usually ending in .py. Importing makes names from that module available under a namespace. Explicit imports make the origin of names clear.

```python
import json
import math
from pathlib import Path

print(math.sqrt(81))
print(json.dumps({"active": True}))
print(Path("notes").name)
```

## 2. Write and import a small module

This example writes a tiny module into a practice folder, adds that folder to this notebook session's module search path, and imports the function.

```python
from pathlib import Path
import sys

module_dir = Path("day3_module_practice")
module_dir.mkdir(exist_ok=True)
(module_dir / "greetings.py").write_text(
    "def greet(name):\n    return f'Hello, {name}!'\n",
    encoding="utf-8",
)
sys.path.insert(0, str(module_dir.resolve()))

import greetings
print(greetings.greet("Learner"))
```

## 3. The __main__ guard

When a file is run as the entry point, its __name__ is "__main__". When imported, the module gets its module name instead. The guard keeps script-only behavior from running during import.

```python
def main():
    print("Run the program")

if __name__ == "__main__":
    main()
```

Why is this guard useful when a module can be both run and imported?

## 4. Packages: folders of modules

A package organizes modules in a directory. A regular package commonly contains __init__.py.

    orders_app/
        __init__.py
        validation.py
        convert.py

orders_app/validation.py is imported as orders_app.validation. Inside a package, a relative import such as from .validation import validate refers to a sibling module. Python also supports namespace packages; use the regular package pattern for this exercise.
