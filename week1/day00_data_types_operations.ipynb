## 2. Numbers: `int`, `float`, and `complex`

Integers and floats support arithmetic. `/` always performs true division; `//` is floor division; `%` is the remainder; `**` is exponentiation. `divmod(a, b)` returns the quotient and remainder together. Complex numbers use `j` for the imaginary part.

````python
a = 17
b = 5
print(a + b, a - b, a * b)
print(a / b, a // b, a % b, a ** 2)
print(divmod(a, b))
print(round(3.14159, 2), abs(-8))
print(2 + 3j, (2 + 3j).real, (2 + 3j).imag)
````

### Numeric conversions and common cautions

`int`, `float`, and `complex` convert compatible values. Converting a float to `int` truncates toward zero. Floating-point values approximate many decimal fractions, so avoid using plain `float` for exact money calculations.

````python
print(int(4.9))
print(float(7))
print(complex(2, 4))
print(0.1 + 0.2)
print(10 ** 100)  # Python integers can grow beyond machine-word size
````

## 3. Booleans, comparisons, and truth values

`bool` has two values: `True` and `False`. Comparisons produce booleans. `and`, `or`, and `not` combine conditions. In conditions, empty containers, zero, `None`, and `False` are false-like; most other values are true-like. `and` and `or` short-circuit and return one of their operands.

````python
age = 24
has_ticket = True
print(age >= 18)
print(age >= 18 and has_ticket)
print(not has_ticket)

for value in [0, 1, "", "x", [], [0], {}, None]:
    print(repr(value), bool(value))

print("fallback" or "default")
print("" or "default")
````

## 5. Strings (`str`)

Strings are immutable sequences of Unicode text. Indexing gets one character; slicing gets a subsequence. A slice `text[start:stop:step]` includes `start` and stops before `stop`. Common operations include concatenation, repetition, membership, case conversion, trimming, splitting, joining, replacing, and searching.

````python
text = "  Data Engineering  "
print(text[2], text[-1])
print(text[2:6], text[:4], text[-4:])
print(text.strip())
print(text.lower(), text.upper(), text.title())
print("Data" in text)
print("ha" * 3)
print("data" + "base")
````

````python
# ==============================================================================
# MASTERING PYTHON STRING SLICING SYNTAX: string[start:stop:step]
# The "::" notation runs across the entire sequence using a custom "step".
# ==============================================================================

text = "ABCDEFGHI"
print(f"Original String: {repr(text)}\n")

# 1. THE ULTIMATE PALINDROME REVERSE: [::-1]
# Steps backward (-1) from the very end to the very beginning.
reversed_text = text[::-1]
print(f"1. Reversed [::-1]: {repr(reversed_text)}")  # Output: 'IHGFEDCBA'

# 2. EVERY OTHER CHARACTER (EVEN INDEXES): [::2]
# Starts at index 0 and skips forward by 2 elements (0, 2, 4, 6, 8).
even_indexes = text[::2]
print(f"2. Every 2nd Char [::2]: {repr(even_indexes)}")  # Output: 'ACEGI'

# 3. EVERY OTHER CHARACTER (ODD INDEXES): [1::2]
# Forces the start parameter to index 1, then skips forward by 2 elements (1, 3, 5, 7).
odd_indexes = text[1::2]
print(f"3. Odd Index Chars [1::2]: {repr(odd_indexes)}")  # Output: 'BDFH'

# 4. REVERSE AND SKIP CONCURRENTLY: [::-2]
# Moves backward from the end, hopping over every second character.
reverse_skipping = text[::-2]
print(f"4. Reverse Skipping [::-2]: {repr(reverse_skipping)}")  # Output: 'IGECA'

# 5. CUSTOM BOUNDED SLICING WITH STEP: [start:stop:step]
# Extracts from index 1 up to index 7 (excluding 7), grabbing every 2nd character.
bounded_slice = text[1:7:2]
print(f"5. Bounded Skip [1:7:2]: {repr(bounded_slice)}")  # Output: 'BDF'
````

````python
message = "red,green,blue"
parts = message.split(",")
print(parts)
print(" | ".join(parts))
print(message.replace("green", "gold"))
print(message.find("green"), message.startswith("red"), message.endswith("blue"))

user = "Ravi"
items = 3
print(f"{user} has {items} items")
````

Common string methods: `strip`, `split`, `join`, `replace`, `find`, `startswith`, `endswith`, `lower`, `upper`, and `title`. Strings cannot be changed in place; methods return new strings.

## 6. Lists (`list`)

Lists are ordered, mutable sequences. They can contain mixed types, though keeping one kind of item per list is often clearer. Common operations: index, slice, membership, `append`, `extend`, `insert`, `remove`, `pop`, `sort`, `reverse`, and `copy`.

````python
nums = [8, 3, 5]
print(nums[0], nums[-1], nums[1:2])
nums.append(2)
nums.extend([9, 1])
nums.insert(1, 7)
print(nums)
nums.remove(7)
last = nums.pop()
print(nums, 'removed:', last)
nums.sort()
print(nums)
nums.reverse()
print(nums)
````

````python
letters = ["a", "b"]
alias = letters
alias.append("c")
print(letters)  # alias and letters refer to the same list

shallow_copy = letters.copy()
shallow_copy.append("d")
print(letters, shallow_copy)

squares = [n * n for n in range(5)]
print(squares)
````

**Key distinction:** `list.sort()` changes the existing list and returns `None`. `sorted(list)` creates a new sorted list. Assignment such as `alias = letters` does not copy a list.

## 7. Tuples (`tuple`)

Tuples are ordered, immutable sequences. Use them for a fixed group of values or to return several values together. A one-item tuple needs a trailing comma: `(5,)`. A tuple may contain a mutable object such as a list; the tuple slot cannot be replaced, but that inner list can still be mutated.

````python
point = (3, 4)
print(point[0], point[-1], point[::-1])
print(point.count(3), point.index(4))
x, y = point
print(x, y)
single_item = (5,)
print(single_item, type(single_item))

record = ("Asha", [10, 20])
record[1].append(30)
print(record)
````

## 8. Ranges (`range`)

A `range` represents a sequence of integers without building a list of all of them at once. It is commonly used in loops. `range(stop)`, `range(start, stop)`, and `range(start, stop, step)` are available; the stop is excluded.

````python
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(10, 0, -2)))
print(3 in range(1, 8, 2))

for index in range(3):
    print(index)
````

## 9. Dictionaries (`dict`)

Dictionaries are mutable mappings from unique, hashable keys to values. They preserve insertion order. Common operations: create, retrieve by key, safe retrieval with `get`, add/update, membership in keys, iterate through `.items()`, and remove with `pop`.

````python
person = {"name": "Maya", "age": 30}
print(person["name"])
print(person.get("city", "unknown"))
person["age"] = 31
person["city"] = "Pune"
person.update({"active": True})
print("age" in person)
for key, value in person.items():
    print(key, value)
city = person.pop("city")
print(city, person)
````

````python
prices = {"tea": 20, "coffee": 35}
discounted = {item: price * 0.9 for item, price in prices.items()}
print(discounted)

counts = {}
for letter in "banana":
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
````

**Dictionary tip:** `person[key]` raises `KeyError` if the key is absent; `person.get(key, default)` safely supplies a default. Dictionary keys must be hashable; lists and dictionaries cannot be keys.

## 10. Sets (`set`) and frozen sets (`frozenset`)

A set contains unique, hashable elements and is useful for membership checks and set algebra. Sets are unordered: don't rely on their display order. Use `add`, `update`, `discard`, and `remove`. `frozenset` is an immutable set and can itself be used as a dictionary key.

````python
a = {1, 2, 2, 3}
b = {3, 4, 5}
print(a)
a.add(6)
a.update([7, 8])
a.discard(99)  # no error if missing
print(2 in a)
print(a | b)  # union
print(a & b)  # intersection
print(a - b)  # in a, not b
print(a ^ b)  # in exactly one set

fixed = frozenset({"read", "write"})
permissions = {fixed: "editor"}
print(permissions[fixed])
````

`remove(value)` raises `KeyError` when absent; `discard(value)` does not. A set cannot contain a list or dictionary because those types are mutable and unhashable.

## 11. Bytes and bytearray

`bytes` is an immutable sequence of byte values (0–255), often used for files and network data. `bytearray` is the mutable counterpart. Text and bytes are different: encode text into bytes, and decode bytes back into text.

````python
text = "café"
encoded = text.encode("utf-8")
print(encoded)
print(encoded.decode("utf-8"))

data = bytearray(b"ABC")
data[0] = ord("Z")
data.extend(b"D")
print(data)
````

## 12. Common operations across types

These built-ins work with many different types: `len`, `min`, `max`, `sum` (numeric iterables), `sorted`, `enumerate`, `zip`, and `reversed`. The operation is only valid when it makes sense for the values involved.

````python
numbers = [4, 1, 9]
print(len(numbers), min(numbers), max(numbers), sum(numbers))
print(sorted(numbers), list(reversed(numbers)))

for index, value in enumerate(["red", "blue"]):
    print(index, value)

names = ["Asha", "Ben"]
scores = [90, 85]
print(list(zip(names, scores)))
````

### Conversions at a glance

Common constructors convert values when possible: `int`, `float`, `str`, `list`, `tuple`, `dict`, `set`, `bytes`, and `bool`. Conversion may fail if the source isn't compatible. Some conversions consume an iterable.

````python
print(int("12"), float("2.5"), str(99))
print(list((1, 2)), tuple([1, 2]), set([1, 1, 2]))
print(dict([("a", 1), ("b", 2)]))
print(bool(""), bool("ok"))
````

## 13. Mutability and identity recap

Immutable built-ins include numbers, booleans, strings, tuples (their slots cannot be reassigned), ranges, bytes, and frozensets. Mutable built-ins include lists, dictionaries, sets, and bytearrays. A tuple can contain a mutable object. `==` checks value equality; `is` checks whether two names refer to the exact same object. Use `is` for `None`, not general value comparison.

````python
first = [1, 2]
second = first
third = first.copy()
print(first == second, first is second)
print(first == third, first is third)
second.append(3)
print(first, second, third)
````
