## 3. Write tests for validation

The example function converts an amount and returns either cleaned data or a reason. Read its behavior, then consider tests for valid, missing, malformed, and negative amounts.

```python
def validate_amount(row):
    try:
        amount = float(row["amount"])
    except (KeyError, ValueError):
        return None, "amount is missing or not numeric"

    if amount < 0:
        return None, "amount must be non-negative"

    return {**row, "amount": amount}, None
```

A test should check the returned cleaned value and the reason. Example for a valid case:

    from your_module import validate_amount

    def test_valid_amount():
        clean, reason = validate_amount({"amount": "4.5"})
        assert clean == {"amount": 4.5}
        assert reason is None

## 4. Parametrize similar cases

pytest.mark.parametrize runs one test function with several input/expected-value pairs. It helps when cases share the same test logic.

    import pytest

    @pytest.mark.parametrize(
        "text, expected",
        [("2", 2), ("0", 0), ("-3", -3)],
    )
    def test_int_conversion(text, expected):
        assert int(text) == expected

## 5. Assert an exception

Use pytest.raises when a call should raise an exception:
