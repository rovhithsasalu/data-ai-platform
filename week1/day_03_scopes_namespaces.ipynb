## 1. Namespace and local scope

A namespace maps names to objects. A scope is the region where a name can be looked up directly. A function call creates a local namespace.

```python
message = "module value"

def show():
    message = "local value"
    print(message)

show()
print(message)
```

## 2. LEGB lookup

When Python looks up a name inside a function, it searches Local, Enclosing function scopes, Global module scope, then Built-ins. It uses the first matching binding.

```python
message = "global"

def outer():
    message = "enclosing"

    def inner():
        print(message)

    inner()

outer()
```

## 3. Local assignment and shadowing

An assignment inside a function normally binds a local name. If Python determines that a name is local, reading it before its local assignment can raise UnboundLocalError—even if a global with the same spelling exists.

```python
value = 10

def read_then_assign():
    # Uncomment to observe UnboundLocalError.
    # print(value)
    value = 20
    return value

print(read_then_assign())
print(value)
```

## 4. global

global declares that assignments to the named variable rebind the module-level name. It affects rebinding, not the ability to mutate a referenced list or dictionary.

```python
status = "new"

def mark_done():
    global status
    status = "done"

mark_done()
print(status)
```

## 5. nonlocal

nonlocal declares that assignments to the named variable rebind a name in the nearest enclosing function scope. It does not target the module global.

```python
def make_counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment

counter = make_counter()
print(counter())
print(counter())
```
